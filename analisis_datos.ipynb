## Introducción

En esta actividad se trabaja con un conjunto de datos obtenido de Hugging Face. La idea es aprender a cargar información, hacer algunos cambios y guardar un resumen de los resultados.

Para esto se crearon dos clases en Python. La primera permite obtener los datos y organizarlos en un DataFrame, mientras que la segunda genera un archivo de texto con información importante sobre esos datos.

Por medio de Jupyter Notebook se conectan las dos clases y se realizan varias modificaciones a la información antes de generar el archivo `resumen.txt`.

## 1. Carga del conjunto de datos

Primero se importan las dos clases creadas en los archivos de Python. Después, se utiliza `HFDatasetLoader` para descargar el conjunto de datos Iris desde Hugging Face y convertirlo en una tabla de Pandas.

Al final, se muestran las primeras cinco filas y la cantidad de filas y columnas que tiene el conjunto de datos al inicio. Esto permite conocer cómo viene organizada la información antes de hacer los cambios.

```python
from cargador_dataset import HFDatasetLoader
from generador_resumen import GeneradorResumen

# Cargar el conjunto de datos desde Hugging Face
cargador = HFDatasetLoader("scikit-learn/iris")
df = cargador.cargar_datos()

# Mostrar las primeras filas
display(df.head())

# Consultar las dimensiones originales
print("Dimensiones originales:", df.shape)
```

## 2. Transformaciones de los datos

Después de cargar los datos, se realizan cuatro cambios para trabajar con la información:

**1. Cambiar el nombre de una columna:** se cambia `SepalLengthCm` por `LongitudSepalo` para que el nombre sea más fácil de identificar.

**2. Filtrar los datos:** se dejan únicamente las flores cuya longitud del pétalo sea mayor que 1.5.

**3. Agregar una columna:** se crea `AreaPetalo`, multiplicando la longitud del pétalo por su anchura.

**4. Seleccionar las columnas:** se dejan seis columnas para el resultado final, incluyendo la nueva que se creó.

Con estos cambios se obtiene `df_transformado`, que es la tabla que se utiliza para generar el resumen.

```python

# Transformación 1: Renombrar una columna
df = df.rename(columns={
    "SepalLengthCm": "LongitudSepalo"
})

# Transformación 2: Filtrar registros
df = df[df["PetalLengthCm"] > 1.5].copy()

# Transformación 3: Crear una nueva columna calculada
df["AreaPetalo"] = df["PetalLengthCm"] * df["PetalWidthCm"]

# Transformación 4: Seleccionar las columnas finales
df_transformado = df[
    [
        "LongitudSepalo",
        "SepalWidthCm",
        "PetalLengthCm",
        "PetalWidthCm",
        "AreaPetalo",
        "Species"
    ]
].copy()

# Mostrar los resultados
display(df_transformado.head())

print("Dimensiones después de las transformaciones:", df_transformado.shape)

```

## 3. Generación del resumen

Cuando terminan los cambios, se utiliza la clase `GeneradorResumen` para crear el archivo `resumen.txt` a partir de `df_transformado`. El archivo contiene la cantidad de filas y columnas, los nombres de las columnas, sus tipos de datos, las primeras cinco filas y la cantidad de valores faltantes, de esta manera, se puede revisar cómo quedó la información después de las transformaciones sin tener que mirar toda la tabla.

```python

# Generar el resumen de los datos transformados
generador = GeneradorResumen(df_transformado)
generador.generar_resumen("resumen.txt")

print("Proceso completado correctamente.")
```
